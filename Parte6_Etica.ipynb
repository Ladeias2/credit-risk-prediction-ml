## Conclusão

Modelos de crédito devem ser desenvolvidos com atenção aos riscos de discriminação e aos impactos das decisões automatizadas sobre os clientes. Mesmo que aumentassem a precisão do modelo, atributos sensíveis como raça, religião, gênero e orientação sexual não deveriam ser utilizados. Além disso, é importante monitorar possíveis variáveis que funcionem como proxies desses atributos. A LGPD exige transparência, finalidade legítima, segurança dos dados e garante ao cidadão o direito de solicitar revisão de decisões automatizadas. Portanto, o modelo deve atuar como apoio à análise humana, promovendo decisões mais justas, responsáveis e alinhadas às exigências legais.
