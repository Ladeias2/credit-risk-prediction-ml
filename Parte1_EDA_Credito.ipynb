```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/credito.csv")

df.head()

print("Dimensões:")
print(df.shape)

print("\nTipos de dados:")
print(df.dtypes)

print("\nResumo:")
display(df.describe())

proporcao = df["inadimplente"].value_counts(normalize=True)

print(proporcao)

print(
    f"\nPercentual de inadimplentes: "
    f"{proporcao[1]*100:.2f}%"
)

```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="inadimplente"
)

plt.title("Distribuição da inadimplência")
plt.xlabel("Inadimplente")
plt.ylabel("Quantidade")

plt.show()
```

```python
ausentes = df.isna().sum()

print(ausentes)

print("\nTotal de ausentes:")
print(ausentes.sum())
```

```python
plt.figure(figsize=(8,4))

ausentes.sort_values(
    ascending=False
).plot(kind="bar")

plt.title("Valores ausentes por coluna")
plt.ylabel("Quantidade")
plt.show()
```

```python
df.groupby(
    "inadimplente"
)["score_credito"].describe()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="inadimplente",
    y="score_credito"
)

plt.title("Score de Crédito por Situação")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.barplot(
    data=df,
    x="possui_imovel",
    y="inadimplente"
)

plt.title("Taxa de Inadimplência por Posse de Imóvel")

plt.show()
```

```python
inad_imovel = (
    df.groupby("possui_imovel")
      ["inadimplente"]
      .mean()
)

display(inad_imovel)
```

```python
plt.figure(figsize=(6,4))

sns.barplot(
    data=df,
    x="possui_imovel",
    y="inadimplente"
)

plt.title("Taxa de Inadimplência por Posse de Imóvel")

plt.show()
```

## Conclusão

A análise exploratória mostrou que aproximadamente 21% dos contratos resultaram em inadimplência, indicando um leve desbalanceamento entre as classes. Foram identificados valores ausentes nas variáveis renda_mensal e tempo_emprego_anos, que deverão ser tratados durante o pré-processamento. Também foi observado que clientes com menor score de crédito apresentam maior probabilidade de inadimplência, enquanto clientes com imóvel tendem a apresentar menor risco. Além disso, a finalidade do empréstimo demonstrou influência no comportamento de pagamento, com destaque para empréstimos destinados a negócios. A variável id_contrato foi descartada por ser apenas um identificador sem valor preditivo para o modelo.
